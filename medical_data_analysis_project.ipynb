# Medical Health Data Analysis Using Python
**Academic project | Synthetic teaching dataset**

> All records in this notebook are simulated for educational demonstration. They do not represent real patients and must not be used for diagnosis, treatment, screening, or clinical decisions.

## Aim
To demonstrate data cleaning, descriptive statistics, grouping, visualization, and correlation analysis using Python.

## 1. Import required libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the accompanying synthetic dataset
df = pd.read_csv("synthetic_medical_health_dataset.csv")
df.head()
```

## 2. Dataset overview

```python
print("Shape:", df.shape)
print("\\nColumn information:")
print(df.info())
print("\\nMissing values:")
print(df.isna().sum())
print("\\nSummary statistics:")
display(df.describe(include="all"))
```

## 3. Data preparation

```python
# Remove duplicate records, if any
df = df.drop_duplicates()

# Check missing values and data types
print("Duplicates remaining:", df.duplicated().sum())
print(df.dtypes)

# BMI grouping is included in the source dataset for demonstration
df[["Patient_ID", "BMI", "BMI_Group"]].head()
```

## 4. Descriptive analysis

```python
print("Average age:", round(df["Age_years"].mean(), 2))
print("Average BMI:", round(df["BMI"].mean(), 2))
print("Average systolic BP:", round(df["Systolic_BP_mmHg"].mean(), 2))
print("Average glucose:", round(df["Glucose_mg_dL"].mean(), 2))
print("\\nBMI groups:")
print(df["BMI_Group"].value_counts())
```

## Age distribution

```python
plt.hist(df['Age_years'], bins=12, edgecolor='black')\nplt.title('Age Distribution')\nplt.xlabel('Age (years)'); plt.ylabel('Count'); plt.show()
```

## Gender distribution

```python
df['Gender'].value_counts().plot(kind='bar', edgecolor='black')\nplt.title('Gender Distribution')\nplt.xlabel('Gender'); plt.ylabel('Count'); plt.xticks(rotation=0); plt.show()
```

## BMI group distribution

```python
df['BMI_Group'].value_counts().plot(kind='bar', edgecolor='black')\nplt.title('BMI Groups')\nplt.xlabel('BMI group'); plt.ylabel('Count'); plt.xticks(rotation=15); plt.show()
```

## Age and systolic BP

```python
plt.scatter(df['Age_years'], df['Systolic_BP_mmHg'], alpha=.6)\nplt.title('Age vs Systolic BP')\nplt.xlabel('Age'); plt.ylabel('Systolic BP (mmHg)'); plt.show()
```

## BMI and glucose

```python
plt.scatter(df['BMI'], df['Glucose_mg_dL'], alpha=.6)\nplt.title('BMI vs Glucose')\nplt.xlabel('BMI'); plt.ylabel('Glucose (mg/dL)'); plt.show()
```

## Cholesterol by gender

```python
df.boxplot(column='Total_Cholesterol_mg_dL', by='Gender')\nplt.title('Cholesterol by Gender'); plt.suptitle(''); plt.ylabel('mg/dL'); plt.show()
```

## Correlation analysis

```python
corr = df[['Age_years','BMI','Systolic_BP_mmHg','Diastolic_BP_mmHg','Glucose_mg_dL','Total_Cholesterol_mg_dL']].corr()\nprint(corr.round(2))\nplt.imshow(corr, vmin=-1, vmax=1); plt.colorbar(); plt.title('Correlation Matrix'); plt.xticks(range(len(corr)), corr.columns, rotation=45, ha='right'); plt.yticks(range(len(corr)), corr.columns); plt.tight_layout(); plt.show()
```

## 5. Findings and conclusion
Review the charts and descriptive statistics. Correlation describes association in this simulated sample; it does not establish causation. The dataset is synthetic and is not suitable for clinical interpretation.
